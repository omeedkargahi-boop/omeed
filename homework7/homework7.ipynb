```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df2 = sns.load_dataset("car_crashes")
df2.head()
```

```python
df2 = df2.dropna()
df2
```

```python
print(df2.shape)
# There are 51 rows and 8 columns
display(df2.head())
display(df2.tail())
df2.columns
# "Total" represents the total number of drivers in fatal collisions*
# "Speeding" represents drivers involved in fatal collisions who were speeding*
# "Alchohol" represents number of drivers involved in fata collisions who were under the influence*
# "Not_distracted" represents the number of drivers involved in collisions who were not distracted*
# "No_previous" represents the number of drivers invovled in fatal collisions who had not been involved in any previous accidents*
# "Ins_premium" represents the average car insurance premium per insured vehicle in U.S. dollars
# "Ins_losses" represents the money lost by insurance companies per insured driver in the U.S.
# "Abbrev" is the abbreviation for the state (Including D.C.)

# *per billion miles driven
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 14))

sns.scatterplot(data = df2, x = "speeding", y = "not_distracted", ax=axes[0,0], color="blue")
axes[0,0].set_title("Speeding vs Not Distracted")
axes[0,0].set_xlabel("Speeding drivers in fatal collisions (per billion miles driven)")
axes[0,0].set_ylabel("Not-distracted drivers in fatal collisions (per billion miles driven)")

sns.scatterplot(data = df2, x = "alcohol", y = "no_previous", ax=axes[1,0], color="red")
axes[1,0].set_title("DUI vs No Previous Accidents")
axes[1,0].set_xlabel("Under-the-influence drivers in fatal collisions (per billion miles driven)")
axes[1,0].set_ylabel("First-time accident drivers in fatal collisions (per billion miles driven)")

sns.histplot(data = df2, x = "ins_premium", ax=axes[0,1], color="green")
axes[0,1].set_title("Insurance Premiums")
axes[0,1].set_xlabel("Average car insurance premium per insured vehicle (USD)")
axes[0,1].set_ylabel("Number of States")

sns.boxplot(data = df2, x = "ins_losses", ax=axes[1,1], color="purple")
axes[1,1].set_title("Distribution of insurance losses")
axes[1,1].set_xlabel("Money lost by insurance companies per insured driver in the U.S. (USD)")
```

```python
# Subplot one represents how speeding grows with not being distracred. As speeding grows, so does not being distracted. So I guess this teaches us that people who speed are usually aware of what they are doing.
# Subplot two represents how average car insurance is in all the states. Most states have a premium of 700-900 USD.
# Subplot three represents how first-time accidents grow with DUI's. I guess this means that drunk drivers are usually first time collisioners. IDK if you ask me it seems off.
# Subplot four shows the distribution of money lost by insurance companies per insuranced driver. It tells us that the minimum out of a state is about $80, the maximum is about $195, and the median is almost $140.
```
