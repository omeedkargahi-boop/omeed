```python
# ---- PART 2: MATPLOTLIB ----

import numpy as np
import matplotlib.pyplot as plt
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1) # Importing needed packages and data
```

```python
# 1
data.shape # Prints shape of array
# There are only 2642 rows!
```

```python
# 3
magnitude = data[:, 0] # Creating magnitude varibale from the initial data
temperature = data[:, 1] # Creating temperature varibale from the initial data

# 4
print(np.min(magnitude)) # This corresponds to brighter stars
print(np.max(magnitude)) # This corresponds to dimmer stars

# 5
print(np.min(temperature)) # This corresponds to cooler stars
print(np.max(temperature)) # This corresponds to hotter stars

# 6
# The star with a lower magnitude is brighter, while the star with a higher magnitude is dimmer
```

```python
# 1
# Creating new variables with the natural logarithm
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

# 2
plt.figure(figsize=(8,6)) # Creating the figure

# 3
plt.scatter(x = log_temperature, y = log_magnitude, label = "General Stars") # Scattering the data

# 4
# Inverts the axes
plt.gca().invert_xaxis() 
plt.gca().invert_yaxis()

# 5
# Uses numpy to find indices of stars with highest and lowest magnitude and temperature
brightest = (np.argmin(magnitude))
dimmest = (np.argmax(magnitude))
coldest = (np.argmin(temperature))
hottest = (np.argmax(temperature))

# Scatters each of the above stars with a unique color, and a star shape rather than the default circle
plt.scatter(x = log_temperature[brightest], y = log_magnitude[brightest], color = "black", marker = "*", s = 300, label = "Brightest Star")
plt.scatter(x = log_temperature[dimmest], y = log_magnitude[dimmest], color = "purple", marker = "*", s = 300, label = "Dimmest Star")
plt.scatter(x = log_temperature[coldest], y = log_magnitude[coldest], color = "green", marker = "*", s = 300, label = "Coldest Star")
plt.scatter(x = log_temperature[hottest], y = log_magnitude[hottest], color = "orange", marker = "*", s = 300, label = "Hottest Star")

# 6
plt.legend() # Creates a legend

# 7, 8, 9
# Titles and labels
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")

plt.show()
```
