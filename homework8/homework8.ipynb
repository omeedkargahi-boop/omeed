```python
import numpy as np
import matplotlib.pyplot as plt
# Importing needed packages
```

```python
def make_sine_wave(x, A, w): # A function desiged to create a sin graph with inputted values
    y = A*np.sin(w*x)
    return y

x = np.linspace(0, 2*np.pi, 1000) # Creates an array that goes from 0 to 2pi with 1000 points, acting as out x-inputs

amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])


fig, axes = plt.subplots(1, 3, figsize=(25,5)) # Plots our three subplots with the correct figsize
zip(amplitude, frequency) # Loops through both arrays

for amp in amplitude: # Plots first subplot with varying amplitudes
    axes[0].plot(x, make_sine_wave(x, amp, 1))

for freq in amplitude: # Plots second subplot with varying frequencies
    axes[1].plot(x, make_sine_wave(x, 1, freq))

for amp, freq in zip(amplitude, frequency): # Plots third subplot with varying amplitudes and frequencies
    axes[2].plot(x, make_sine_wave(x, amp, freq))

# Creating titles and axes labels for each subplot
axes[0].set_title("Sine Graph with Varying Amplitude", fontsize = 20)
axes[0].set_ylabel("Output Value", fontsize = 18)
axes[0].set_xlabel("Angle(radians)", fontsize = 18)

axes[1].set_title("Sine Graph with Varying Frequency", fontsize = 20)
axes[1].set_ylabel("Output Value", fontsize = 18)
axes[1].set_xlabel("Angle(radians)", fontsize = 18)

axes[2].set_title("Sine Graph with Varying Amplitude & Frequency", fontsize = 20)
axes[2].set_ylabel("Output Value", fontsize = 18)
axes[2].set_xlabel("Angle(radians)", fontsize = 18)

plt.show() # Displays the plots
```
